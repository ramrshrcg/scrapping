```python
import csv
import pandas as pd
import os
from pathlib import Path

def fix_csv_columns(input_file, output_file):
    """
    Fix CSV by ensuring only two columns regardless of quotes in content
    """
    print(f"Processing: {input_file}")
    
    fixed_rows = []
    
    # Read the original CSV line by line
    with open(input_file, 'r', encoding='utf-8') as f:
        lines = f.readlines()
    
    print(f"  Total lines to process: {len(lines)}")
    
    # Process each line
    for i, line in enumerate(lines):
        line = line.strip()
        if not line:
            continue
            
        # For the header row, just use it as is
        if i == 0:
            fixed_rows.append(['शीर्षक', 'विवरण'])
            continue
        
        # Find the first comma - this should be the separator between title and description
        first_comma = line.find(',')
        
        if first_comma == -1:
            # No comma found, skip this line or handle as error
            print(f"  Warning: No comma found in line {i+1}, skipping")
            continue
        
        # Split into title and description
        title = line[:first_comma].strip()
        description = line[first_comma+1:].strip()
        
        # Remove surrounding quotes from title and description if they exist
        if title.startswith('"') and title.endswith('"'):
            title = title[1:-1]
        if description.startswith('"') and description.endswith('"'):
            description = description[1:-1]
        
        # Replace any escaped quotes with regular quotes
        title = title.replace('""', '"')
        description = description.replace('""', '"')
        
        # Add to fixed rows
        fixed_rows.append([title, description])
    
    # Write the fixed CSV
    with open(output_file, 'w', encoding='utf-8', newline='') as f:
        writer = csv.writer(f, quoting=csv.QUOTE_ALL)
        for row in fixed_rows:
            writer.writerow(row)
    
    print(f"  ✅ Fixed CSV saved to: {output_file}")
    print(f"  Total rows processed: {len(fixed_rows)}")
    
    return len(fixed_rows)

def find_csv_files(folder_path):
    """
    Recursively find all CSV files in the given folder and subfolders
    """
    csv_files = []
    folder_path = Path(folder_path)
    
    # Use glob to find all CSV files recursively
    csv_files = list(folder_path.rglob("*.csv"))
    
    return csv_files

def fix_all_csv_files(folder_path):
    """
    Find and fix all CSV files in the folder and its subfolders
    """
    print(f"🔍 Searching for CSV files in: {folder_path}")
    
    # Find all CSV files
    csv_files = find_csv_files(folder_path)
    
    if not csv_files:
        print("❌ No CSV files found in the specified folder and its subfolders.")
        return
    
    print(f"📁 Found {len(csv_files)} CSV files to process:")
    for file in csv_files:
        print(f"  - {file}")
    
    print("\n" + "="*60)
    print("🚀 Starting processing...")
    print("="*60)
    
    # Process each CSV file
    processed_count = 0
    total_rows_processed = 0
    failed_files = []
    
    for csv_file in csv_files:
        try:
            # Create output filename by adding '_fixed' before the extension
            input_path = Path(csv_file)
            output_path = input_path
            
            # Skip if it's already a fixed file
            if "_fixed" in input_path.stem:
                print(f"⏭️  Skipping already fixed file: {csv_file}")
                continue
            
            print(f"\n📄 Processing file {processed_count + 1}:")
            rows_processed = fix_csv_columns(str(csv_file), str(output_path))
            
            processed_count += 1
            total_rows_processed += rows_processed
            
        except Exception as e:
            print(f"❌ Error processing {csv_file}: {e}")
            failed_files.append(str(csv_file))
    
    # Summary
    print("\n" + "="*60)
    print("📊 PROCESSING SUMMARY")
    print("="*60)
    print(f"✅ Successfully processed: {processed_count} files")
    print(f"📈 Total rows processed: {total_rows_processed}")
    
    if failed_files:
        print(f"❌ Failed files: {len(failed_files)}")
        for failed_file in failed_files:
            print(f"  - {failed_file}")
    else:
        print("🎉 All files processed successfully!")

# Set your folder path here
folder_path = '/Users/ramchandraghmire/Desktop/Datas/scrapping/News_csv'

```

```python
try:
    fix_all_csv_files(folder_path)
except Exception as e:
    print(f"❌ Error: {e}")
    import traceback
    traceback.print_exc()
```
