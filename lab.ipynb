```python
from pathlib import Path
import random

file_path = Path("unique_nepali_sentences.txt")
sample_size = 10

if not file_path.exists():
    raise FileNotFoundError(f"File not found: {file_path.resolve()}")

# Reservoir sampling: uniform random sample from a huge file using O(sample_size) memory.
reservoir = []
non_empty_count = 0

with file_path.open("r", encoding="utf-8", errors="ignore") as f:
    for raw_line in f:
        line = raw_line.strip()
        if not line:
            continue

        non_empty_count += 1

        if len(reservoir) < sample_size:
            reservoir.append(line)
        else:
            j = random.randint(1, non_empty_count)
            if j <= sample_size:
                reservoir[j - 1] = line

print(f"Total non-empty lines scanned: {non_empty_count}")
print(f"Sample size returned: {len(reservoir)}")
print("\nRandom sample lines:")
for i, line in enumerate(reservoir, start=1):
    print(f"{i}. {line}")
```

```python
from pathlib import Path
import random

file_path = Path("unique_nepali_sentences.txt")
chunk_size = 15  # Number of consecutive non-empty lines to print

if not file_path.exists():
    raise FileNotFoundError(f"File not found: {file_path.resolve()}")

# Pass 1: count non-empty lines (memory-safe)
non_empty_total = 0
with file_path.open("r", encoding="utf-8", errors="ignore") as f:
    for raw_line in f:
        if raw_line.strip():
            non_empty_total += 1

if non_empty_total == 0:
    raise ValueError("The file has no non-empty lines.")

chunk_size = min(chunk_size, non_empty_total)
max_start = non_empty_total - chunk_size + 1
start_line_index = 1
# random.randint(1, max_start)  # 1-based among non-empty lines
end_line_index = start_line_index + chunk_size - 1

# Pass 2: stream and collect only the chosen consecutive chunk
chunk = []
current_non_empty_index = 0

with file_path.open("r", encoding="utf-8", errors="ignore") as f:
    for raw_line in f:
        line = raw_line.strip()
        if not line:
            continue

        current_non_empty_index += 1

        if current_non_empty_index < start_line_index:
            continue
        if current_non_empty_index > end_line_index:
            break

        chunk.append(line)

print(f"Total non-empty lines: {non_empty_total}")
print(f"Chunk range (non-empty line index): {start_line_index} to {end_line_index}")
print("\nConsecutive lines from one chunk:")
for offset, line in enumerate(chunk, start=0):
    print(f"{line}")
```
