```python
import pandas as pd
import os
import glob
import re
from pathlib import Path

def merge_csv_files_nepali_only():
    """
    Merge all CSV files from eKantipur directories, remove ALL line breaks 
    and filter to keep only Nepali (Devanagari) characters
    """
    # /Users/ramchandraghmire/Desktop/Datas/scrapping/AnnapurnaPost
    # Define base paths
    base_path = "/Users/ramchandraghmire/Desktop/Datas/scrapping/"
    # ekantipur_path = os.path.join(base_path, "eKantipur")
    past_news_path = os.path.join(base_path, "News_csv")
    
    # Store all content as one continuous string
    all_content = []
    seen_sentences = set()
    
    print("🔍 Starting to merge AnnapurnaPost CSV files (Nepali characters only)...")
    # print(f"📁 eKantipur path: {ekantipur_path}")
    print(f"📁 Past news path: {past_news_path}")
    
    def clean_nepali_text(text):
        """Clean text to keep only Nepali (Devanagari) characters and basic punctuation"""
        if not text or pd.isna(text):
            return ""
        
        text = str(text).strip()
        
        # Remove ALL newline characters first
        text = re.sub(r'\n+', ' ', text)
        
        # Keep only Devanagari characters (U+0900-U+097F), numbers, spaces, and basic punctuation
        # Devanagari range includes Nepali characters
        cleaned_text = re.sub(r'[^\u0900-\u097F\u0020\u002C\u002E\u003A\u003B\u003F\u0021\u0028\u0029\u002D\u0030-\u0039]+', '', text)
        
        # Clean up multiple spaces
        cleaned_text = re.sub(r'\s+', ' ', cleaned_text).strip()
        
        return cleaned_text
    
    def clean_and_add_content(title, description):
        """Clean content by removing non-Nepali characters and add if not already seen"""
        title = clean_nepali_text(title)
        description = clean_nepali_text(description)
        
        # Add title if unique and not empty
        if title and title not in seen_sentences:
            all_content.append(title)
            seen_sentences.add(title)
            
        # Add description if unique and not empty
        if description and description not in seen_sentences:
            all_content.append(description)
            seen_sentences.add(description)
    
    # # Process files from eKantipur directory (direct CSV files)
    # if os.path.exists(ekantipur_path):
    #     csv_files = glob.glob(os.path.join(ekantipur_path, "*.csv"))
    #     print(f"\n📊 Found {len(csv_files)} CSV files in eKantipur directory")
        
    #     for file_path in csv_files:
    #         print(f"   Processing: {os.path.basename(file_path)}")
    #         try:
    #             df = pd.read_csv(file_path)
    #             # Process each row: add title then description to continuous flow
    #             for _, row in df.iterrows():
    #                 clean_and_add_content(row.iloc[0], row.iloc[1])
    #         except Exception as e:
    #             print(f"   ❌ Error processing {file_path}: {e}")
    
    # Process files from ekantipur_past_news subdirectories
    if os.path.exists(past_news_path):
        # Find all CSV files in all subdirectories
        csv_pattern = os.path.join(past_news_path, "**", "*.csv")
        csv_files = glob.glob(csv_pattern, recursive=True)
        print(f"\n📊 Found {len(csv_files)} CSV files in past news subdirectories")
        
        for file_path in csv_files:
            relative_path = os.path.relpath(file_path, past_news_path)
            print(f"   Processing: {relative_path}")
            try:
                df = pd.read_csv(file_path)
                # Process each row: add title then description to continuous flow
                for _, row in df.iterrows():
                    clean_and_add_content(row.iloc[0], row.iloc[1])
            except Exception as e:
                print(f"   ❌ Error processing {file_path}: {e}")
    
    # Save to text file as ONE CONTINUOUS LINE with only Nepali characters
    output_file = "merged_News_csv_.txt"
    output_path = os.path.join("/Users/ramchandraghmire/Desktop/Datas/scrapping", output_file)
    
    with open(output_path, 'w', encoding='utf-8') as f:
        # Join all content with spaces, ensure only Nepali characters
        continuous_text = ' '.join(all_content)
        # Final cleanup: ensure no remaining non-Nepali characters
        continuous_text = re.sub(r'[^\u0900-\u097F\u0020\u002C\u002E\u003A\u003B\u003F\u0021\u0028\u0029\u002D\u0030-\u0039]+', '', continuous_text)
        
        # Add proper spacing after punctuation marks
        # Add space after purnabiram (।) if not already present
        continuous_text = re.sub(r'।(?!\s)', '। ', continuous_text)
        # Add space after period (.) if not already present
        continuous_text = re.sub(r'\.(?!\s)', '. ', continuous_text)
        # Add space after comma (,) if not already present  
        continuous_text = re.sub(r',(?!\s)', ', ', continuous_text)
        # Add space after colon (:) if not already present
        continuous_text = re.sub(r':(?!\s)', ': ', continuous_text)
        # Add space after question mark (?) if not already present
        continuous_text = re.sub(r'\?(?!\s)', '? ', continuous_text)
        # Add space after exclamation mark (!) if not already present
        continuous_text = re.sub(r'!(?!\s)', '! ', continuous_text)
        
        # Clean up any multiple spaces that might have been created
        continuous_text = re.sub(r'\s+', ' ', continuous_text).strip()
        
        # Write as one continuous line
        f.write(continuous_text)
    
    print(f"\n✅ Successfully merged all CSV files!")
    print(f"📄 Total unique content pieces: {len(all_content):,}")
    print(f"💾 Output saved to: {output_path}")
    print(f"📝 Format: ALL content in ONE continuous line, ONLY Nepali characters")
    
    return output_path, len(all_content)

# Run the Nepali-only merge function
output_file_nepali, total_nepali = merge_csv_files_nepali_only()
```
